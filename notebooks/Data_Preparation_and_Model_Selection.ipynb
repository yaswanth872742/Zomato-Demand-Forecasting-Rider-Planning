```python
# Checking the missing values
missing_values = data1_clean.isnull().sum()

missing_percentage = (missing_values / len(data1_clean)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing %": missing_percentage.round(2)
})

missing_summary = missing_summary[missing_summary["Missing Values"] > 0]

missing_summary.sort_values("Missing Values", ascending=False)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

missing_summary["Missing Values"].plot(kind="bar")

plt.title("Missing Values in DATA1")
plt.ylabel("Count")
plt.xlabel("Columns")

plt.xticks(rotation=45)

plt.show()
```

```python
data1_clean_v2 = data1_clean.copy()

# Filling categorical missing values
data1_clean_v2["type"] = data1_clean_v2["type"].fillna("Unknown")

data1_clean_v2["timings"] = data1_clean_v2["timings"].fillna("Not Available")

data1_clean_v2["cuisines"] = data1_clean_v2["cuisines"].fillna("Unknown")

data1_clean_v2["address"] = data1_clean_v2["address"].fillna("Unknown")

data1_clean_v2["opentable_support"] = data1_clean_v2["opentable_support"].fillna(
    data1_clean_v2["opentable_support"].mode()[0]
)
```

```python
daily_demand = (
    eda_data.groupby("Order Date")
    .size()
    .reset_index(name="Orders")
)

daily_demand.head()
```

```python
print(daily_demand.shape)

daily_demand.describe()
```

```python
plt.figure(figsize=(15,5))

plt.plot(
    daily_demand["Order Date"],
    daily_demand["Orders"],
    linewidth=2
)

plt.title(
    "Daily Total Orders",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Date")
plt.ylabel("Orders")

plt.grid(alpha=0.4)

plt.tight_layout()

plt.show()
```

```python
daily_demand["Year"] = daily_demand["Order Date"].dt.year
daily_demand["Month"] = daily_demand["Order Date"].dt.month
daily_demand["Day"] = daily_demand["Order Date"].dt.day
daily_demand["DayOfWeek"] = daily_demand["Order Date"].dt.dayofweek
daily_demand["Quarter"] = daily_demand["Order Date"].dt.quarter
daily_demand["IsWeekend"] = (
    daily_demand["DayOfWeek"] >= 5
).astype(int)
```

```python
daily_demand["Lag_1"] = daily_demand["Orders"].shift(1)

daily_demand["Lag_7"] = daily_demand["Orders"].shift(7)

daily_demand["Lag_30"] = daily_demand["Orders"].shift(30)
```

```python
daily_demand["Rolling_Mean_7"] = (
    daily_demand["Orders"]
    .rolling(7)
    .mean()
)

daily_demand["Rolling_STD_7"] = (
    daily_demand["Orders"]
    .rolling(7)
    .std()
)
```

```python
daily_demand = daily_demand.dropna().reset_index(drop=True)
```

```python
daily_demand.head()

daily_demand.info()

daily_demand.describe()
```

# Modeling

```python
features = [
    "Year",
    "Month",
    "Day",
    "DayOfWeek",
    "Quarter",
    "IsWeekend",
    "Lag_1",
    "Lag_7",
    "Lag_30",
    "Rolling_Mean_7",
    "Rolling_STD_7"
]
```

```python
X = daily_demand[features]

y = daily_demand["Orders"]
```

```python
split = int(len(daily_demand) * 0.8)

X_train = X.iloc[:split]
X_test = X.iloc[split:]

y_train = y.iloc[:split]
y_test = y.iloc[split:]
```

```python
print("Training Samples :", len(X_train))
print("Testing Samples :", len(X_test))
```

```python
baseline_pred = X_test["Lag_1"]

baseline_mae = np.mean(np.abs(y_test - baseline_pred))

print("Baseline MAE:", baseline_mae)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np
import pandas as pd
```

```python
# Model creating and training
lr_model = LinearRegression()

lr_model.fit(X_train, y_train)

lr_pred = lr_model.predict(X_test)
```

```python
lr_mae = mean_absolute_error(y_test, lr_pred)

lr_rmse = np.sqrt(mean_squared_error(y_test, lr_pred))

lr_r2 = r2_score(y_test, lr_pred)

print("Linear Regression Performance")
print("--------------------------------")
print("MAE :", round(lr_mae,3))
print("RMSE:", round(lr_rmse,3))
print("R²  :", round(lr_r2,3))
```

```python
plt.figure(figsize=(15,5))

plt.plot(
    y_test.values,
    label="Actual",
    linewidth=2
)

plt.plot(
    lr_pred,
    label="Predicted",
    linewidth=2
)

plt.title(
    "Linear Regression\nActual vs Predicted",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Time")

plt.ylabel("Orders")

plt.legend()

plt.grid(alpha=0.4)

plt.tight_layout()

plt.show()
```

```python
from xgboost import XGBRegressor
```

```python
xgb_model.fit(X_train, y_train)
```

```python
xgb_pred = xgb_model.predict(X_test)
```

```python
xgb_mae = mean_absolute_error(y_test, xgb_pred)

xgb_rmse = np.sqrt(mean_squared_error(y_test, xgb_pred))

xgb_r2 = r2_score(y_test, xgb_pred)

print("XGBoost Performance")
print("------------------------------")
print("MAE :", round(xgb_mae,3))
print("RMSE:", round(xgb_rmse,3))
print("R²  :", round(xgb_r2,3))
```

```python
plt.figure(figsize=(15,5))

plt.plot(y_test.values, label="Actual", linewidth=2)

plt.plot(xgb_pred, label="Predicted", linewidth=2)

plt.title(
    "XGBoost\nActual vs Predicted",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Time")
plt.ylabel("Orders")

plt.legend()

plt.grid(alpha=0.4)

plt.tight_layout()

plt.show()
```
